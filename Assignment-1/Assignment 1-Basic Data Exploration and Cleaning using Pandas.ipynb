-Handle missing values (identify, fill/drop)

```python
df.isnull()
```

```python
print(df.isnull().sum())
```

```python
The columns named discount, what_customers_said, seller_name, videos, seller_information, variations contains missing values.
```

```python
-Iterate through each column
```

```python
for col in df.columns:
    if df[col].isna().any():#if column has missing values
        if pd.api.types.is_numeric_dtype(df[col]): # if col is numeric
            df[col]=df[col].fillna(df[col].median()) # filling the missing values with median
        else:
            mode=df[col].mode(dropna=True)
            fill_value = mode.iloc[0] if not mode.empty else 'Unknown'
            df[col] = df[col].fillna(fill_value)
    
```

```python
print(df.isnull().sum().sum())
```

Filtering rows where first numeric column is grater than its median


```python
numeric_cols=df.select_dtypes(include="number").columns
first_col=numeric_cols[0]
```

```python
middle_value=df[first_col].median()
high_values=df[df[first_col]>middle_value]

print(f"Rows where {first_col} is greater than {middle_value}:")
print(high_values.head())
```

-Remove Duplicates

```python
print("Rows before removing duplicates", len(df))
df=df.drop_duplicates()
print("Rows after removing duplicates", len(df))
```

-Create a derived column (total_amount = price * quantity)

```python
df['quantity'] = 1
df['final_price'] = df['final_price'].astype(str).str.replace('₹', '').str.replace(',', '').astype(float)
df['total_amount'] = df['final_price'] * df['quantity']
print(df[['final_price', 'quantity', 'total_amount']].head())
```
